```python
# Import necessary libraries for visualization
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set style for better-looking plots
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

print("Visualization libraries imported successfully!")
print(f"Matplotlib version: {plt.matplotlib.__version__}")
print(f"Seaborn version: {sns.__version__}")
```

```python
# Load the cleaned dataset
df = pd.read_csv('student_scores_cleaned.csv')

print("Dataset loaded successfully!")
print(f"\nDataset shape: {df.shape}")
print("\nFirst few rows:")
print(df.head())
print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Calculate average scores for each subject
print("CALCULATING AVERAGE SCORES")
print("="*50)

# Calculate average score for Math
avg_math = df['math_score'].mean()
print(f"\nAverage Math Score: {avg_math:.2f}")

# Calculate average score for Science
avg_science = df['science_score'].mean()
print(f"Average Science Score: {avg_science:.2f}")

# Calculate average score for English
avg_english = df['english_score'].mean()
print(f"Average English Score: {avg_english:.2f}")

print("\n" + "="*50)
print("\nSummary of Average Scores:")
print(f"Math: {avg_math:.2f}")
print(f"Science: {avg_science:.2f}")
print(f"English: {avg_english:.2f}")
```

```python
#  Prepare data for bar chart (average scores)
subjects = ['Math', 'Science', 'English']
average_scores = [avg_math, avg_science, avg_english]

# Create a DataFrame for better visualization
scores_df = pd.DataFrame({
    'Subject': subjects,
    'Average Score': average_scores
})

print("Data prepared for Bar Chart:")
print("="*50)
print(scores_df)
print("\nData is ready for visualization!")
```

```python
# Count number of students by gender
print("COUNTING STUDENTS BY GENDER")
print("="*50)

# Count students by gender
gender_counts = df['gender'].value_counts()
print("\nStudent count by gender:")
print(gender_counts)

print("\n" + "="*50)
# Calculate percentages
total_students = len(df)
print(f"\nTotal students: {total_students}")
print("\nPercentage distribution:")
for gender, count in gender_counts.items():
    percentage = (count / total_students) * 100
    print(f"{gender}: {count} students ({percentage:.1f}%)")
```

```python
# Create bar chart for average scores per subject
plt.figure(figsize=(10, 6))

# Create bar chart using matplotlib
bars = plt.bar(subjects, average_scores, color=['#3498db', '#2ecc71', '#e74c3c'],
               edgecolor='black', linewidth=1.2)

# Add title and labels
plt.title('Average Scores per Subject', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Subject', fontsize=12, fontweight='bold')
plt.ylabel('Average Score', fontsize=12, fontweight='bold')

# Add value labels on top of bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'{height:.2f}',
             ha='center', va='bottom', fontsize=11, fontweight='bold')

# Add grid for better readability
plt.grid(axis='y', alpha=0.3, linestyle='--')
plt.ylim(0, 100)

# Display the chart
plt.tight_layout()
plt.show()

print("Bar chart created successfully!")
```

```python
# Create pie chart for number of students by gender
plt.figure(figsize=(8, 8))

# Define colors for pie chart
colors = ['#3498db', '#e74c3c']

# Create pie chart
plt.pie(gender_counts.values,
        labels=gender_counts.index,
        autopct='%1.1f%%',
        colors=colors,
        startangle=90,
        explode=(0.05, 0.05),  # Slightly separate slices
        shadow=True,
        textprops={'fontsize': 12, 'fontweight': 'bold'})

# Add title
plt.title('Number of Students by Gender', fontsize=16, fontweight='bold', pad=20)

# Equal aspect ratio ensures pie chart is circular
plt.axis('equal')

# Display the chart
plt.tight_layout()
plt.show()

print("Pie chart created successfully!")
```

```python
#  Create an alternative bar chart using Seaborn
plt.figure(figsize=(10, 6))

# Create bar chart using seaborn
sns.barplot(x='Subject', y='Average Score', data=scores_df,
            palette=['#3498db', '#2ecc71', '#e74c3c'],
            edgecolor='black', linewidth=1.5)

# Add title and labels
plt.title('Average Scores per Subject (Seaborn)', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Subject', fontsize=12, fontweight='bold')
plt.ylabel('Average Score', fontsize=12, fontweight='bold')

# Add value labels on bars
for i, score in enumerate(average_scores):
    plt.text(i, score + 1, f'{score:.2f}',
             ha='center', fontsize=11, fontweight='bold')

plt.ylim(0, 100)
plt.tight_layout()
plt.show()

print("Seaborn bar chart created successfully!")
```

```python
# Create both charts in a single figure (side by side)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Bar Chart on the left
bars = ax1.bar(subjects, average_scores, color=['#3498db', '#2ecc71', '#e74c3c'],
               edgecolor='black', linewidth=1.2)
ax1.set_title('Average Scores per Subject', fontsize=14, fontweight='bold', pad=15)
ax1.set_xlabel('Subject', fontsize=11, fontweight='bold')
ax1.set_ylabel('Average Score', fontsize=11, fontweight='bold')
ax1.grid(axis='y', alpha=0.3, linestyle='--')
ax1.set_ylim(0, 100)

# Add value labels on bars
for bar in bars:
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height,
             f'{height:.2f}', ha='center', va='bottom',
             fontsize=10, fontweight='bold')

# Pie Chart on the right
colors = ['#3498db', '#e74c3c']
ax2.pie(gender_counts.values, labels=gender_counts.index,
        autopct='%1.1f%%', colors=colors, startangle=90,
        explode=(0.05, 0.05), shadow=True,
        textprops={'fontsize': 11, 'fontweight': 'bold'})
ax2.set_title('Number of Students by Gender', fontsize=14, fontweight='bold', pad=15)

plt.tight_layout()
plt.show()

print("Combined visualization created successfully!")
```
