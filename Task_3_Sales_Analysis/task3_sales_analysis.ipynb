```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set plotting style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

print("Libraries imported successfully!")
print(f"Pandas version: {pd.__version__}")
print(f"NumPy version: {np.__version__}")
print(f"Matplotlib version: {plt.matplotlib.__version__}")
print(f"Seaborn version: {sns.__version__}")
```

```python
#  Load the dataset using Pandas
df = pd.read_csv('tips.csv')

print("Dataset loaded successfully!")
print("="*60)
print(f"\nDataset Name: Tips Dataset")
print(f"Total Records: {len(df)}")
print(f"Total Columns: {len(df.columns)}")
print("\n" + "="*60)
```

```python
#  Show the first 5 rows of the dataset
print("FIRST 5 ROWS OF THE DATASET")
print("="*60)
print(df.head())
print("\n" + "="*60)
print("First 5 rows displayed successfully!")
```

```python
# Display column names
print("COLUMN NAMES IN THE DATASET")
print("="*60)

columns = df.columns.tolist()
print(f"\nTotal Columns: {len(columns)}")
print("\nColumn List:")
for i, col in enumerate(columns, 1):
    print(f"{i}. {col}")

print("\n" + "="*60)
```

```python
#  Display dataset shape
print("DATASET SHAPE")
print("="*60)

shape = df.shape
print(f"\nShape: {shape}")
print(f"Number of Rows: {shape[0]}")
print(f"Number of Columns: {shape[1]}")
print(f"\nThis dataset contains {shape[0]} records and {shape[1]} features.")
print("\n" + "="*60)
```

```python
#  Display data types and dataset information
print("DATASET INFORMATION")
print("="*60)
print("\nData Types and Non-Null Counts:")
print(df.info())

print("\n" + "="*60)
print("\nData Types Summary:")
print(df.dtypes)
```

```python
# Show basic statistics using describe()
print("BASIC STATISTICAL SUMMARY")
print("="*60)
print("\nDescriptive Statistics for Numeric Columns:")
print(df.describe())

print("\n" + "="*60)
print("\nDescriptive statistics displayed successfully!")
```

```python
# Calculate total bills by gender (similar to Product A and B)
print("BASIC CALCULATIONS - TOTAL BILLS BY GENDER")
print("="*60)

# Calculate total bills for each gender (treating as Product A and B)
total_male = df[df['sex'] == 'Male']['total_bill'].sum()
total_female = df[df['sex'] == 'Female']['total_bill'].sum()

print(f"\nTotal Bills for Male customers: ${total_male:.2f}")
print(f"Total Bills for Female customers: ${total_female:.2f}")
print(f"\nOverall Total Bills: ${total_male + total_female:.2f}")

print("\n" + "="*60)
```

```python
# Calculate average tip (treating tip as profit)
print("AVERAGE TIP CALCULATION (Profit)")
print("="*60)

# Calculate average tip
avg_tip = df['tip'].mean()
total_tip = df['tip'].sum()
median_tip = df['tip'].median()

print(f"\nAverage Tip (Profit): ${avg_tip:.2f}")
print(f"Median Tip: ${median_tip:.2f}")
print(f"Total Tips: ${total_tip:.2f}")

print("\n" + "="*60)
print(f"On average, customers tip ${avg_tip:.2f} per meal.")
```

```python
#  Calculate maximum and minimum bills for each gender
print("MAXIMUM AND MINIMUM BILLS BY GENDER")
print("="*60)

# For Male customers
male_max = df[df['sex'] == 'Male']['total_bill'].max()
male_min = df[df['sex'] == 'Male']['total_bill'].min()

print("\nMale Customers:")
print(f"  Maximum Bill: ${male_max:.2f}")
print(f"  Minimum Bill: ${male_min:.2f}")

# For Female customers
female_max = df[df['sex'] == 'Female']['total_bill'].max()
female_min = df[df['sex'] == 'Female']['total_bill'].min()

print("\nFemale Customers:")
print(f"  Maximum Bill: ${female_max:.2f}")
print(f"  Minimum Bill: ${female_min:.2f}")

print("\n" + "="*60)
```

```python
# Prepare data for visualizations
print("PREPARING DATA FOR VISUALIZATION")
print("="*60)

# Data for bar chart - Total bills by gender
gender_totals = df.groupby('sex')['total_bill'].sum().reset_index()
gender_totals.columns = ['Gender', 'Total_Bills']

print("\nTotal Bills by Gender:")
print(gender_totals)

# Data for line chart - Average tip by day
day_order = ['Thur', 'Fri', 'Sat', 'Sun']
tip_by_day = df.groupby('day')['tip'].mean().reindex(day_order).reset_index()
tip_by_day.columns = ['Day', 'Average_Tip']

print("\nAverage Tip by Day (Profit Trend):")
print(tip_by_day)

print("\n" + "="*60)
print("Data prepared for visualization!")
```

```python
#  Create bar chart for total bills by gender
plt.figure(figsize=(10, 6))

# Create bar chart
bars = plt.bar(gender_totals['Gender'], gender_totals['Total_Bills'],
               color=['#3498db', '#e74c3c'],
               edgecolor='black', linewidth=1.5, width=0.6)

# Add title and labels
plt.title('Total Bills by Gender', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Gender', fontsize=12, fontweight='bold')
plt.ylabel('Total Bills ($)', fontsize=12, fontweight='bold')

# Add value labels on top of bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'${height:.2f}',
             ha='center', va='bottom', fontsize=11, fontweight='bold')

# Add grid
plt.grid(axis='y', alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()

print("📊 Bar Chart Description:")
print("This bar chart shows the total bills by gender, with male customers")
print("generating significantly higher total bills than female customers.")
```

```python
# Create line chart for tip trend over days
plt.figure(figsize=(10, 6))

# Create line chart
plt.plot(tip_by_day['Day'], tip_by_day['Average_Tip'],
         marker='o', linewidth=2.5, markersize=10,
         color='#2ecc71', markerfacecolor='#e74c3c',
         markeredgecolor='black', markeredgewidth=1.5)

# Add title and labels
plt.title('Average Tip Trend Over Days (Profit Trend)',
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Day of Week', fontsize=12, fontweight='bold')
plt.ylabel('Average Tip ($)', fontsize=12, fontweight='bold')

# Add value labels on data points
for i, row in tip_by_day.iterrows():
    plt.text(row['Day'], row['Average_Tip'] + 0.05,
             f'${row["Average_Tip"]:.2f}',
             ha='center', fontsize=10, fontweight='bold')

# Add grid
plt.grid(axis='both', alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()

print("📈 Line Chart Description:")
print("This line chart displays the average tip (profit) trend across different")
print("days of the week, showing how tipping patterns vary from Thursday to Sunday.")
```

```python
#  Create both charts in a single figure
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Bar Chart on the left
bars = ax1.bar(gender_totals['Gender'], gender_totals['Total_Bills'],
               color=['#3498db', '#e74c3c'],
               edgecolor='black', linewidth=1.5, width=0.6)
ax1.set_title('Total Bills by Gender', fontsize=14, fontweight='bold', pad=15)
ax1.set_xlabel('Gender', fontsize=11, fontweight='bold')
ax1.set_ylabel('Total Bills ($)', fontsize=11, fontweight='bold')
ax1.grid(axis='y', alpha=0.3, linestyle='--')

for bar in bars:
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height,
             f'${height:.2f}', ha='center', va='bottom',
             fontsize=10, fontweight='bold')

# Line Chart on the right
ax2.plot(tip_by_day['Day'], tip_by_day['Average_Tip'],
         marker='o', linewidth=2.5, markersize=10,
         color='#2ecc71', markerfacecolor='#e74c3c',
         markeredgecolor='black', markeredgewidth=1.5)
ax2.set_title('Average Tip Trend Over Days', fontsize=14, fontweight='bold', pad=15)
ax2.set_xlabel('Day of Week', fontsize=11, fontweight='bold')
ax2.set_ylabel('Average Tip ($)', fontsize=11, fontweight='bold')
ax2.grid(axis='both', alpha=0.3, linestyle='--')

for i, row in tip_by_day.iterrows():
    ax2.text(row['Day'], row['Average_Tip'] + 0.05,
             f'${row["Average_Tip"]:.2f}',
             ha='center', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()

print("Combined visualization created successfully!")
```

```python
#  Additional exploratory analysis
print("ADDITIONAL EXPLORATORY ANALYSIS")
print("="*60)

# Analyze by day
print("\nTotal Bills by Day:")
print(df.groupby('day')['total_bill'].sum().sort_values(ascending=False))

print("\n" + "-"*60)

# Analyze by time
print("\nTotal Bills by Time (Lunch vs Dinner):")
print(df.groupby('time')['total_bill'].sum())

print("\n" + "-"*60)

# Analyze by smoker status
print("\nTotal Bills by Smoker Status:")
print(df.groupby('smoker')['total_bill'].sum())

print("\n" + "="*60)
```

```python
# Analyze correlation between bill and tip
print("CORRELATION ANALYSIS")
print("="*60)

# Calculate correlation
correlation = df['total_bill'].corr(df['tip'])
print(f"\nCorrelation between Total Bill and Tip: {correlation:.4f}")

# Create scatter plot to visualize relationship
plt.figure(figsize=(10, 6))
plt.scatter(df['total_bill'], df['tip'], alpha=0.6,
            color='#3498db', edgecolor='black', s=50)

plt.title('Relationship between Total Bill and Tip',
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Total Bill ($)', fontsize=12, fontweight='bold')
plt.ylabel('Tip ($)', fontsize=12, fontweight='bold')
plt.grid(alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()

print(f"\nThere is a strong positive correlation ({correlation:.2f}) between")
print("total bill and tip amount, meaning higher bills tend to have higher tips.")
```
