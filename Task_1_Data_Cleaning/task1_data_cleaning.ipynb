```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully!")
print(f"Pandas version: {pd.__version__}")
print(f"NumPy version: {np.__version__}")
```

```python
df = pd.read_csv('student_scores.csv')

print("Dataset loaded successfully!")
print(f"\nDataset shape: {df.shape}")
print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")
print("\nFirst few rows of the dataset:")
df.head()
```

```python
print("Dataset Information:")
print("="*50)
df.info()

print("\n" + "="*50)
print("\nColumn Names:")
print(df.columns.tolist())

print("\n" + "="*50)
print("\nBasic Statistics:")
df.describe()
```

```python
# Check for missing values in the dataset
print("Missing Values Analysis:")
print("="*50)

missing_values = df.isnull().sum()
print("\nMissing values per column:")
print(missing_values)

print("\n" + "="*50)
missing_percentage = (df.isnull().sum() / len(df)) * 100
print("\nMissing values percentage:")
print(missing_percentage)

print("\n" + "="*50)
print(f"\nTotal missing values in dataset: {df.isnull().sum().sum()}")
```

```python
# Clean column names (remove extra spaces, make consistent)
print("Original column names:")
print(df.columns.tolist())

# Strip whitespace and make consistent
df.columns = df.columns.str.strip()
df.columns = df.columns.str.replace(' ', '_')
df.columns = df.columns.str.lower()

print("\n" + "="*50)
print("\nCleaned column names:")
print(df.columns.tolist())
print("\nColumn names cleaned successfully!")
```

```python
# Handle missing values - Fill with column mean for numeric columns
print("Handling missing values...")
print("="*50)

# Store original shape
original_shape = df.shape

# Fill missing numeric values with column mean
numeric_columns = df.select_dtypes(include=[np.number]).columns
for col in numeric_columns:
    if df[col].isnull().sum() > 0:
        mean_value = df[col].mean()
        df[col].fillna(mean_value, inplace=True)
        print(f"Filled missing values in '{col}' with mean: {mean_value:.2f}")

print("\n" + "="*50)
print(f"\nOriginal shape: {original_shape}")
print(f"Current shape: {df.shape}")
print(f"\nRemaining missing values: {df.isnull().sum().sum()}")
```

```python
# Check for duplicate records
print("Checking for duplicate records...")
print("="*50)

duplicate_count = df.duplicated().sum()
print(f"\nNumber of duplicate rows: {duplicate_count}")

if duplicate_count > 0:
    print("\nDuplicate rows found:")
    print(df[df.duplicated(keep=False)])
else:
    print("\nNo duplicate rows found in the dataset.")
```

```python
# Remove duplicate records if any exist
print("Removing duplicate records...")
print("="*50)

shape_before = df.shape
df.drop_duplicates(inplace=True)
shape_after = df.shape

rows_removed = shape_before[0] - shape_after[0]
print(f"\nShape before removing duplicates: {shape_before}")
print(f"Shape after removing duplicates: {shape_after}")
print(f"Number of duplicate rows removed: {rows_removed}")

# Reset index after removing duplicates
df.reset_index(drop=True, inplace=True)
print("\nIndex reset successfully!")
```

```python
#  Display cleaned dataset summary
print("CLEANED DATASET SUMMARY")
print("="*50)

print(f"\nFinal shape: {df.shape}")
print(f"Total students: {len(df)}")
print(f"Total missing values: {df.isnull().sum().sum()}")

print("\n" + "="*50)
print("\nCleaned column names:")
for i, col in enumerate(df.columns, 1):
    print(f"{i}. {col}")

print("\n" + "="*50)
print("\nFirst 5 rows of cleaned data:")
df.head()
```

```python
# Perform basic statistical analysis
print("BASIC STATISTICAL ANALYSIS")
print("="*50)

print("\nDescriptive Statistics:")
print(df.describe())

print("\n" + "="*50)
print("\nGender Distribution:")
print(df['gender'].value_counts())

print("\n" + "="*50)
print("\nAge Distribution:")
print(df['age'].value_counts().sort_index())
```

```python
# Analyze scores by subject
print("SUBJECT-WISE SCORE ANALYSIS")
print("="*50)

print("\nMath Scores:")
print(f"Mean: {df['math_score'].mean():.2f}")
print(f"Median: {df['math_score'].median():.2f}")
print(f"Min: {df['math_score'].min():.2f}")
print(f"Max: {df['math_score'].max():.2f}")

print("\n" + "-"*50)
print("\nScience Scores:")
print(f"Mean: {df['science_score'].mean():.2f}")
print(f"Median: {df['science_score'].median():.2f}")
print(f"Min: {df['science_score'].min():.2f}")
print(f"Max: {df['science_score'].max():.2f}")

print("\n" + "-"*50)
print("\nEnglish Scores:")
print(f"Mean: {df['english_score'].mean():.2f}")
print(f"Median: {df['english_score'].median():.2f}")
print(f"Min: {df['english_score'].min():.2f}")
print(f"Max: {df['english_score'].max():.2f}")
```

```python
# Save the cleaned dataset
output_filename = 'student_scores_cleaned.csv'
df.to_csv(output_filename, index=False)

print("TASK COMPLETED!")
print("="*50)
print(f"\nCleaned dataset saved as: {output_filename}")
print(f"Total records: {len(df)}")
print(f"Total columns: {len(df.columns)}")
print("\nData cleaning and basic analysis completed successfully!")
```
